```
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

import warnings
warnings.filterwarnings("ignore")
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style="white", color_codes=True)

iris_data = load_iris()
iris = pd.DataFrame(data = iris_data['data'], columns = iris_data['feature_names'])

iris.head()
```

```
iris.plot(kind="scatter", x="sepal length (cm)", y="sepal width (cm)")
plt.show()
```

```
print(iris.columns)
iris['species'] = iris_data.target
#iris['species'] = iris['target'].apply(lambda x: iris.target[x])
print(iris.columns)
print(iris.head())
```

```
iris.plot(kind="scatter", x="sepal length (cm)", y="sepal width (cm)",color="green",s=70)
plt.show()
```

```
sns.jointplot(x="sepal length (cm)", y="sepal width (cm)", data=iris, size=5)
plt.show()
```

```
plt.figure(figsize=(8, 6))
for species in iris['species'].unique():
subset = iris[iris['species'] == species]
plt.scatter(subset['sepal length (cm)'], subset['sepal width (cm)'], label=species)
plt.title('Sepal Length vs Sepal Width in Iris Dataset', fontsize=16)
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.legend(title="Species", fontsize=10)
plt.grid(True)
plt.show()
```

```
sns.pairplot(iris, hue='species', diag_kind='kde', palette='Set2')
```

```
avg_sepal_length = iris.groupby('species')['sepal length (cm)'].mean()
# Plot bar chart
plt.figure(figsize=(8, 6))
avg_sepal_length.plot(kind='bar', color=['skyblue', 'orange', 'green'])
plt.title('Average Sepal Length by Species', fontsize=16)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Average Sepal Length (cm)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```
species_count = iris['species'].value_counts()
# Plot pie chart
plt.figure(figsize=(8, 6))
species_count.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['lightcoral', 'lightblue', 'lightgreen'])
plt.title('Iris Species Distribution', fontsize=16)
plt.ylabel('')  # Remove default ylabel
plt.show()
```

```
# Calculate average dimensions by species
avg_dimensions = iris.groupby('species')[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']].mean()
# Plot stacked bar chart
avg_dimensions.plot(kind='bar', stacked=True, figsize=(10, 6), color=['skyblue', 'orange', 'green', 'red'])
plt.title('Average Dimensions by Species', fontsize=16)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Average Dimension (cm)', fontsize=12)
plt.legend(title='Features', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```
# Calculate average petal width
avg_petal_width = iris.groupby('species')['petal width (cm)'].mean()
# Plot horizontal bar chart
plt.figure(figsize=(8, 6))
avg_petal_width.plot(kind='barh', color=['salmon', 'gold', 'lightgreen'])
plt.title('Average Petal Width by Species', fontsize=16)
plt.xlabel('Average Petal Width (cm)', fontsize=12)
plt.ylabel('Species', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
